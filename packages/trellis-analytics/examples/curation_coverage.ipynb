# Curation coverage

Host 2 of 3 for `trellis_analytics.routines.curation_coverage` (brief 2b). Run it with papermill, `slug` and `ods_url` as parameters. It prints what the routine returns; nothing is counted here.

```python
slug = "coco_pharma"          # the database slug
ods_url = ""                  # the ODS: sqlite:///... or postgresql://...; empty reads REGISTRY_DATABASE_URL
resource_type = "database"
```

```python
import json, os
from trellis_analytics.readers.ods import OdsReader
from trellis_analytics.routines import curation_coverage

result = curation_coverage(OdsReader(ods_url or os.environ["REGISTRY_DATABASE_URL"]), resource_type, slug)
print("state:", result.envelope.result_state.words, "·", result.envelope.result_state_reason or "no reason needed")
print(result.summary)
for g in result.groups:
    for l in g.lines:
        print(f"{g.key:10} {l.key:20} {l.count:4}  rows {l.rows}")
```
